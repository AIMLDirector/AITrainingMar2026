```python
import matplotlib.pyplot as plt

# 1. Get non-null counts automatically
counts = df_cleaned.count()

# 2. Plotting
plt.figure(figsize=(10, 7))
plt.pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=140)

plt.title('Distribution of Data Presence Across Columns')
plt.axis('equal')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Create a cross-tabulation of Team and Gender
ct = pd.crosstab(df_cleaned['Team'], df_cleaned['Gender'])

# 2. Set up a grid of subplots (e.g., 2 rows, 5 columns if you have 10 teams)
fig, axes = plt.subplots(nrows=2, ncols=5, figsize=(18, 8))
axes = axes.flatten()

# 3. Loop through each team and plot a pie
for i, team in enumerate(ct.index):
    axes[i].pie(ct.loc[team], labels=ct.columns, autopct='%1.1f%%', 
                colors=['#ff9999', '#66b3ff'], startangle=90)
    axes[i].set_title(team)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Aggregate your data
gender_counts = df_cleaned.groupby(['Team', 'Gender']).size().unstack()
plt.figure(figsize=(12, 6))
# 2. Set up the subplot
plt.subplot(1, 2, 1)

# 3. PLOT FIRST (assign to 'ax' to access containers)
ax = gender_counts.plot(kind='bar', ax=plt.gca())

# 4. ADD LABELS AFTER (loop through each gender's bars)
for container in ax.containers:
    ax.bar_label(container, padding=3)

plt.title("Gender Distribution per Team")
plt.tight_layout()
plt.show()
```
